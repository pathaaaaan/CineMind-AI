### Preprocessinig (NLP)

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stopwords = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def PreProcess(text):
  text = str(text).lower()
  text = re.sub(r'[^a-zA-Z\s]','',text)

  words = text.split()
  # remove stopwords
  words = [word for word in words if word not in stopwords]
  # lemmatize
  words = [lemmatizer.lemmatize(word) for word in words]

  return ' '.join(words)
```

```python
# applying preprocessing on 'tags' column
df['tags'] = df['tags'].apply(PreProcess)
```

### Word Embeddings

```python
df = df.reset_index(drop=True)
```

```python
df.rename(columns={'original_title': 'title'}, inplace=True)
```

```python
indices = pd.Series(df.index, index = df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=5000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_mat = tfidf.fit_transform(df['tags'])
```

```python
tfidf_mat
```

### Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
  if title not in indices:
    return ['movie not found!']

  idx = indices[title]
  similarity = cosine_similarity(tfidf_mat[idx], tfidf_mat).flatten()
  similarity_idx = similarity.argsort()[::-1][1:n+1]
  return df.iloc[similarity_idx]['title']
```
